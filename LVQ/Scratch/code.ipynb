```python
import numpy as np
```

```python
X = np.array([[1.0, 1.0],
              [1.5, 2.0],
              [5.0, 5.0],
              [6.0, 5.5]])
y = np.array([0, 0, 1, 1])

alpha = 0.3
epochs = 5
```

```python
classes = np.unique(y)
prototypes = np.array([X[np.where(y == c)[0][0]] for c in classes], dtype=float)
proto_labels = classes.copy()
```

```python
print("Initial prototypes:")
for pl, p in zip(proto_labels, prototypes):
    print(f" Class {pl}: {p}")
```

```python
for epoch in range(1, epochs + 1):
    for xi, yi in zip(X, y):
        dists = np.linalg.norm(prototypes - xi, axis=1)
        winner_idx = np.argmin(dists)
        if proto_labels[winner_idx] == yi:
            prototypes[winner_idx] += alpha * (xi - prototypes[winner_idx])
        else:
            prototypes[winner_idx] -= alpha * (xi - prototypes[winner_idx])
    print(f"\nPrototypes after epoch {epoch}:")
    for pl, p in zip(proto_labels, prototypes):
        print(f" Class {pl}: {p}")
```

```python
final_prototypes = prototypes.copy()
preds = []
for xi in X:
    dists = np.linalg.norm(final_prototypes - xi, axis=1)
    preds.append(proto_labels[np.argmin(dists)])
preds = np.array(preds)
accuracy = (preds == y).mean()
```

```python
print("\nFinal prototypes:")
for pl, p in zip(proto_labels, final_prototypes):
    print(f" Class {pl}: {p}")
print(f"\nTraining accuracy: {accuracy*100:.2f}%")
```
