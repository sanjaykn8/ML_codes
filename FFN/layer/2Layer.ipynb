```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
class TwoLayerForward:
    def __init__(self, W1, b1, W2, b2):
        self.W1, self.b1, self.W2, self.b2 = W1, b1, W2, b2
    def forward(self, x):
        z1 = self.W1.dot(x) + self.b1
        a1 = sigmoid(z1)
        print(f"Hidden net inputs: {z1}")
        print(f"Hidden activations: {a1}")
        z2 = self.W2.dot(a1) + self.b2
        a2 = sigmoid(z2)
        print(f"Output net input: {z2}")
        print(f"Output activation: {a2}")
        return a2
```

```python
if __name__ == '__main__':
    W1 = np.array([[0.1, 0.3], [0.2, 0.4]])
    b1 = np.array([0.1, -0.1])
    W2 = np.array([[0.5, -0.6]])
    b2 = np.array([0.2])
    x = np.array([1, 0])
    
    model = TwoLayerForward(W1, b1, W2, b2)
    output = model.forward(x)
```
